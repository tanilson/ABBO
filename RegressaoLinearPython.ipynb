```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Dicionário de Dados
dados = {
    'horas': [1, 2, 2, 3, 4, 4, 5, 6, 7, 8],
    'nota':  [4, 5, 4.5, 6, 6.5, 7, 7.5, 8, 9, 9.5]
}

df = pd.DataFrame(dados)

# X = entrada
X = df[['horas']]

# y = variável que queremos prever
y = df['nota']

# Treinamento
modelo = LinearRegression()
modelo.fit(X, y)

# Visualização
plt.scatter(df['horas'], df['nota'])
plt.plot(df['horas'], modelo.predict(X))
plt.xlabel('Horas de estudo')
plt.ylabel('Nota')
plt.show()
```
